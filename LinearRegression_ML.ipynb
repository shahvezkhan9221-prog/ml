##Data Imputation

```python
# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns
```

```python
num_cols
```

```python
cat_cols
```

```python
# Impute numerical columns with the mean
num_imputer = SimpleImputer(strategy='mean')
df[num_cols] = num_imputer.fit_transform(df[num_cols])
```

```python
# Impute categorical columns with the mode
cat_imputer = SimpleImputer(strategy='most_frequent')
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

##Data Transfornation : One Hot Encoding

```python
# Apply One-Hot Encoding to categorical columns
# drop_first=True helps avoid the dummy variable trap
df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

```python
df_encoded.head()
```

```python
# Convert boolean values (True/False) to integers (1/0)
df_encoded = df_encoded.astype(int)
# Display the first few rows to verify the change
df_encoded.head()
```

##Feature Scaling : Min-Max Normalization

```python
X = df_encoded.drop('Price', axis=1)
y = df_encoded['Price']
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
# Initialize and apply Min-Max Scaler to the features
scaler = MinMaxScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
X_scaled.head()
```

##Train Test Split

```python
from sklearn.model_selection import train_test_split

# Split data (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

##Model Training

```python
from sklearn.linear_model import LinearRegression

# Initialize the model
model = LinearRegression()
```

```python
# Train the model on the training data
model.fit(X_train, y_train)
```

```python
# Make predictions on the test set
y_pred = model.predict(X_test)
```

```python
y_pred
```

##Regression Evaluation Metrics

```python
# MAE, MSE, RMSE, R2 Score, Ajusted R2 Score
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
print("Mean Absolute Error (MAE):", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error (MSE):", mean_squared_error(y_test, y_pred))
```

```python
print("Root Mean Squared Error (RMSE):", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R-squared (R2) Score:", r2_score(y_test, y_pred))
```
